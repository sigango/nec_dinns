```python
import scipy.io
import pandas as pd

data = scipy.io.loadmat("./boundaries/BC_bello.mat", squeeze_me=True, struct_as_record=False)

ROW_LABELS = [
    "Itot0", "Hi0", "Me0", "Hr0", "Hs0", "He0", "Ms0", "Mi0",
    "beta_h", "beta_m", "gamma_h", "lambda", "mu_h", "mu_m", "theta_h", "theta_m",
]

keys = [k for k in data.keys() if not k.startswith("__")]
print("Keys:", keys)

for k in keys:
    v = data[k]
    print(f"\n--- {k} ---")
    print(f"  type : {type(v)}")
    if hasattr(v, "shape"):
        print(f"  shape: {v.shape}")
    try:
        df = pd.DataFrame(v, index=ROW_LABELS[:len(v)])
        display(df)
    except Exception:
        print(v)
```
